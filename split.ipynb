Splits a track into vocals, drums, bass and other, plus an
instrumental. Free GPU, nothing to install on your phone.

2. Tap ▶ on **Step 2**, choose your file, wait for the ZIP

*RU: Делит трек на вокал, ударные, бас и остальное и собирает
минус. Нажми ▶ на шаге 1, потом на шаге 2 и выбери файл.*

```python
#@title Step 1 · Install (about a minute)
!pip -q install demucs
!wget -q -O split.py https://raw.githubusercontent.com/imBeyondiDentity/Split/main/split.py
import torch
print("ready · GPU:", torch.cuda.is_available())
```

```python
#@title Step 2 · Upload a track and split it
fmt = "mp3" #@param ["mp3", "wav"]
model = "htdemucs" #@param ["htdemucs", "htdemucs_ft"]

import os, shutil
from google.colab import files

uploaded = files.upload()
name = next(iter(uploaded))
base = os.path.splitext(name)[0]

!python split.py "{name}" -f {fmt} -m {model} -o out --overwrite

folder = os.path.join("out", base + "_split")
shutil.make_archive(base + "_split", "zip", folder)
files.download(base + "_split.zip")
```

No GPU? Runtime → Change runtime type → T4 GPU. The CPU works
too, only slower.

*RU: Нет GPU — Runtime → Change runtime type → T4 GPU. На CPU тоже
работает, только медленнее.*
